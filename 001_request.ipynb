```python
# Load env variable
from dotenv import load_dotenv
load_dotenv()
```

```python
# create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5-20251001"
```

```python
# Make a request
message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages = [
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence."
        }
    ]
)
```
